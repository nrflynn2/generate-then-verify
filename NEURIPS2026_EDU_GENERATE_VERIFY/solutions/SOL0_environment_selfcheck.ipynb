<em style="color:#6B6461;font-size:14.5px;">Solutions notebook. Read this if a check failed, or if you want to know why these ten and not others.</em>

X0 has no learner answer, so there is nothing to work through. What follows is the
reasoning behind the ten required checks, which is the part worth reading: each one
guards against a specific failure that shows up in practice, not a hypothetical one.

**Python 3.10 and the path anchor.** The anchor is positional, not absolute: the
root is the parent of `exercises/` or `solutions/`, and `data/` hangs off the root.
That rule survives being unzipped anywhere, being opened from Colab, and being run
headless by a grader. An absolute path in a teaching bundle is a bug that only shows
up on somebody else's machine.

**Checksums over presence.** Checking that a file exists catches a broken download.
Checking its SHA-256 catches the case that happens in practice: a learner opened the
CSV in a spreadsheet, saved it, and now their numbers disagree with the solutions by
a rounding step. The manifest makes that diagnosable in one line.

**The comment='#' negative control.** This is the only check whose passing condition
is that something goes wrong. Passing `comment='#'` to pandas blanks 16 cells in
`boltz2_comparison.csv`, because two of the SMILES in this bundle contain `C#N` and
pandas reads the `#` as the start of a comment. No exception is raised. The table
just quietly loses data. Firing it once here, with the damage counted, is cheaper
than meeting it for the first time in X10 and concluding that Boltz-2 returned nulls.

**The canary.** `0.8089776623453144` is the pairwise diversity of the twenty
DiffSBDD ligands under `Chem.RDKFingerprint` at library defaults, averaged over the
190 unordered pairs, with `Chem.RemoveHs` applied first. Change the fingerprint and
you get a different number; change the hydrogen handling and you get a different
number again. One float therefore tests the RDKit version, the preparation step and
the pair-counting convention at once.

On the shipped SDF the `RemoveHs` call is defensive rather than corrective, because
that file carries no explicit hydrogens: measured with RDKit 2026.03.5, the value is
identical with and without it. Hydrogen handling is decisive in general and inert on
this particular file. The call stays in because a learner who reads the batch with
`removeHs=False`, or who adds hydrogens for a strain calculation and then reuses the
molecules, meets the real version of the trap immediately.

**One deliberate cost.** The canary is the same number X6 asks the learner to
compute, so running X0 hands away half of X6's first part before they reach it. That
was a choice rather than an oversight: a self-check that proves the machine can
produce the exact value a later exercise depends on is worth more than the surprise
of one figure, and X6's graded content is the comparison between two fingerprints and
the sentence a learner would write beside the number, neither of which X0 gives away.
A lecturer who would rather keep the surprise can replace the canary's printed value
with a boolean by editing one line of the check cell.

**Why the optional four are optional.** PoseBusters, PoseCheck and FlowMol3 are the
tools that produced the shipped tables. Reproducing those tables is a module
walkthrough, not an exercise, precisely because it needs installs the CPU-offline
guarantee cannot promise. Every exercise reads the committed result instead. The GPU
line is reported for the same reason a thermometer is reported: to establish that
its absence is not what is going wrong.
